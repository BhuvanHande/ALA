```python
import hashlib
import os
import pickle
import re
from itertools import combinations
from typing import Dict, List, Optional, Set, Tuple

import networkx as nx
import spacy


class KnowledgeGraphBuilder:
    """
    ECAP-RAG graph builder v3.

    Extends the original Week-2 graph builder with:
      - provenance-aware claims
      - document/chunk/source-sentence metadata
      - deterministic claim IDs
      - MultiDiGraph edges so different predicates between the same
        entity pair remain separate
      - explicit co-occurrence edges marked as weak discovery signals
      - compatibility with the original SVO/coplanar extraction logic

    Important:
      `extraction_score` is a heuristic extraction-strength score.
      It is NOT a calibrated probability/confidence.
    """

    TARGET_ENT_TYPES = {
        "PERSON", "ORG", "WORK_OF_ART", "DATE",
        "GPE", "LOC", "NORP", "EVENT", "FAC",
    }

    SUBJ_DEPS = {"nsubj", "nsubjpass"}
    OBJ_DEPS = {"dobj", "attr", "oprd"}
    COMPLEMENT_DEPS = {"oprd", "attr"}
    COPULA_COMPLEMENT_DEPS = {"attr", "acomp"}
    PASSIVE_SUBJ_DEP = "nsubjpass"

    SVO_EDGE_WEIGHT = 1.0
    COOCCUR_EDGE_WEIGHT = 0.25

    def __init__(self, spacy_model: str = "en_core_web_sm"):
        self.nlp = spacy.load(spacy_model)
        self.graph = nx.MultiDiGraph()
        self._alias_index: Dict[str, str] = {}
        self._claim_counter = 0

    # ------------------------------------------------------------------
    # Canonicalization
    # ------------------------------------------------------------------
    def _canonical_key(self, text: str) -> str:
        cleaned = text.strip().lower()
        cleaned = re.sub(r"^(the|a|an)\s+", "", cleaned)
        cleaned = re.sub(r"['\u2019]s\b", "", cleaned)
        cleaned = re.sub(r"[^\w\s.\-]", "", cleaned)
        cleaned = re.sub(r"\s+", " ", cleaned).strip()
        return cleaned

    def _resolve_node_id(self, display_text: str) -> str:
        key = self._canonical_key(display_text)
        if not key:
            key = display_text.strip().lower()
        if key not in self._alias_index:
            self._alias_index[key] = display_text.strip()
        return self._alias_index[key]

    # ------------------------------------------------------------------
    # Sentence-aware chunking
    # ------------------------------------------------------------------
    def chunk_document(
        self, text: str, doc_id: str, chunk_size_words: int = 250
    ) -> List[Tuple[str, str]]:
        doc = self.nlp(text)
        chunks = []
        current_sents: List[str] = []
        current_word_count = 0
        chunk_idx = 0

        for sent in doc.sents:
            sent_text = sent.text.strip()
            if not sent_text:
                continue

            sent_word_count = len(sent_text.split())

            if current_sents and current_word_count + sent_word_count > chunk_size_words:
                chunk_idx += 1
                chunks.append(
                    (f"{doc_id}_chunk_{chunk_idx}", " ".join(current_sents))
                )
                current_sents, current_word_count = [], 0

            current_sents.append(sent_text)
            current_word_count += sent_word_count

        if current_sents:
            chunk_idx += 1
            chunks.append(
                (f"{doc_id}_chunk_{chunk_idx}", " ".join(current_sents))
            )

        return chunks

    # ------------------------------------------------------------------
    # Entity extraction
    # ------------------------------------------------------------------
    def _extract_entities(self, doc):
        return [
            (ent.text, ent.label_, ent.start_char, ent.end_char)
            for ent in doc.ents
            if ent.label_ in self.TARGET_ENT_TYPES
        ]

    # ------------------------------------------------------------------
    # SVO extraction
    # ------------------------------------------------------------------
    def _span_for_token(self, token) -> str:
        for chunk in token.doc.noun_chunks:
            if chunk.start <= token.i < chunk.end:
                return chunk.text

        span_tokens = sorted(
            (
                t
                for t in token.subtree
                if t.dep_ in {"compound", "amod", "poss"} or t.i == token.i
            ),
            key=lambda t: t.i,
        )
        return " ".join(t.text for t in span_tokens) if span_tokens else token.text

    def _expand_conjuncts(self, token):
        result = [token]
        for child in token.children:
            if child.dep_ == "conj" and child.pos_ == token.pos_:
                result.append(child)
        return result

    def _find_agent(self, verb_token):
        for child in verb_token.children:
            if child.dep_ == "agent":
                pobj = next(
                    (gc for gc in child.children if gc.dep_ == "pobj"), None
                )
                if pobj is not None:
                    return pobj
        return None

    def _find_complement(self, verb_token, dep_set: Set[str]):
        return next(
            (c for c in verb_token.children if c.dep_ in dep_set), None
        )

    def _extract_svo_triples(self, doc):
        """
        Returns:
          (subject, relation, object, sentence_text, extraction_score)

        The score is a deterministic heuristic:
          1.00 = direct SVO/canonical copular extraction
          0.90 = passive resultative
          0.85 = prepositional-object fallback
        """
        triples = []

        for sent in doc.sents:
            sent_text = sent.text.strip()

            for token in sent:
                is_copula_root = token.pos_ == "AUX" and token.dep_ == "ROOT"

                if not is_copula_root and token.pos_ != "VERB":
                    continue

                # Copular: X is Y
                if is_copula_root:
                    subj_token = self._find_complement(token, self.SUBJ_DEPS)
                    comp_token = self._find_complement(
                        token, self.COPULA_COMPLEMENT_DEPS
                    )
                    if subj_token is None or comp_token is None:
                        continue

                    for s_tok in self._expand_conjuncts(subj_token):
                        for c_tok in self._expand_conjuncts(comp_token):
                            triples.append(
                                (
                                    self._span_for_token(s_tok),
                                    "is",
                                    self._span_for_token(c_tok),
                                    sent_text,
                                    1.00,
                                )
                            )
                    continue

                is_passive = any(
                    c.dep_ == self.PASSIVE_SUBJ_DEP for c in token.children
                )
                subj_token = self._find_complement(token, self.SUBJ_DEPS)

                if subj_token is None:
                    continue

                if is_passive:
                    agent_token = self._find_agent(token)
                    relation = token.lemma_.lower()

                    if agent_token is not None:
                        for o_tok in self._expand_conjuncts(subj_token):
                            triples.append(
                                (
                                    self._span_for_token(agent_token),
                                    relation,
                                    self._span_for_token(o_tok),
                                    sent_text,
                                    1.00,
                                )
                            )
                        continue

                    complement_token = self._find_complement(
                        token, self.COMPLEMENT_DEPS
                    )
                    if complement_token is not None:
                        for s_tok in self._expand_conjuncts(subj_token):
                            for c_tok in self._expand_conjuncts(
                                complement_token
                            ):
                                triples.append(
                                    (
                                        self._span_for_token(s_tok),
                                        relation,
                                        self._span_for_token(c_tok),
                                        sent_text,
                                        0.90,
                                    )
                                )
                        continue

                    continue

                # Active voice
                obj_token = self._find_complement(token, self.OBJ_DEPS)

                if obj_token is None:
                    for child in token.children:
                        if child.dep_ == "prep":
                            pobj = next(
                                (
                                    gc
                                    for gc in child.children
                                    if gc.dep_ == "pobj"
                                ),
                                None,
                            )
                            if pobj is not None:
                                obj_token = pobj
                                break

                if obj_token is None:
                    continue

                relation = token.lemma_.lower()
                score = 1.00 if obj_token.dep_ in self.OBJ_DEPS else 0.85

                for s_tok in self._expand_conjuncts(subj_token):
                    for o_tok in self._expand_conjuncts(obj_token):
                        triples.append(
                            (
                                self._span_for_token(s_tok),
                                relation,
                                self._span_for_token(o_tok),
                                sent_text,
                                score,
                            )
                        )

        return triples

    # ------------------------------------------------------------------
    # Provenance / claim helpers
    # ------------------------------------------------------------------
    def _make_claim_id(
        self,
        document_id: str,
        chunk_id: str,
        sentence_text: str,
        subject: str,
        relation: str,
        object_: str,
    ) -> str:
        raw = "||".join(
            [
                document_id,
                chunk_id,
                sentence_text,
                subject,
                relation,
                object_,
            ]
        )
        return "claim_" + hashlib.sha1(raw.encode("utf-8")).hexdigest()[:16]

    def _add_claim_to_edge(
        self,
        u: str,
        v: str,
        relation: str,
        chunk_id: str,
        document_id: str,
        sentence_text: str,
        extraction_score: float,
        edge_type: str = "svo",
    ) -> None:
        claim_id = self._make_claim_id(
            document_id,
            chunk_id,
            sentence_text,
            u,
            relation,
            v,
        )

        claim = {
            "claim_id": claim_id,
            "document_id": document_id,
            "chunk_id": chunk_id,
            "source_span": sentence_text,
            "subject": u,
            "relation": relation,
            "object": v,
            "extraction_score": extraction_score,
        }

        # Different predicates stay separate because this is a MultiDiGraph.
        edge_key = relation

        if self.graph.has_edge(u, v, key=edge_key):
            data = self.graph[u][v][edge_key]
            data.setdefault("claims", {})
            data["claims"][claim_id] = claim
            data.setdefault("chunks", set()).add(chunk_id)
            data["weight"] = data.get("weight", 0.0) + (
                self.SVO_EDGE_WEIGHT if edge_type == "svo" else self.COOCCUR_EDGE_WEIGHT
            )
        else:
            self.graph.add_edge(
                u,
                v,
                key=edge_key,
                relation=relation,
                edge_type=edge_type,
                chunks={chunk_id},
                weight=(
                    self.SVO_EDGE_WEIGHT
                    if edge_type == "svo"
                    else self.COOCCUR_EDGE_WEIGHT
                ),
                claims={claim_id: claim},
            )

    # ------------------------------------------------------------------
    # Graph assembly
    # ------------------------------------------------------------------
    def add_chunk(
        self,
        chunk_id: str,
        text: str,
        document_id: Optional[str] = None,
    ) -> None:
        document_id = document_id or chunk_id.split("_chunk_")[0]

        doc = self.nlp(text)
        entities = self._extract_entities(doc)

        for (
            subj_text,
            relation,
            obj_text,
            sentence_text,
            extraction_score,
        ) in self._extract_svo_triples(doc):
            subj_id = self._resolve_node_id(subj_text)
            obj_id = self._resolve_node_id(obj_text)

            if subj_id == obj_id:
                continue

            self._add_or_update_node(
                subj_id,
                chunk_id,
                document_id=document_id,
            )
            self._add_or_update_node(
                obj_id,
                chunk_id,
                document_id=document_id,
            )

            self._add_claim_to_edge(
                subj_id,
                obj_id,
                relation,
                chunk_id,
                document_id,
                sentence_text,
                extraction_score,
                edge_type="svo",
            )

        for text_span, ent_type, start_char, end_char in entities:
            node_id = self._resolve_node_id(text_span)
            self._add_or_update_node(
                node_id,
                chunk_id,
                document_id=document_id,
                ent_type=ent_type,
                mention={
                    "document_id": document_id,
                    "chunk_id": chunk_id,
                    "text": text_span,
                    "start_char": start_char,
                    "end_char": end_char,
                },
            )

        # Weak discovery edges. They are explicitly marked so PathRAG can
        # exclude them from primary reasoning paths by default.
        entity_node_ids = sorted(
            {self._resolve_node_id(t) for t, *_ in entities}
        )

        for a, b in combinations(entity_node_ids, 2):
            if (
                self.graph.has_edge(a, b, key="co_occurs")
                or self.graph.has_edge(b, a, key="co_occurs")
                or self._has_reasoning_edge(a, b)
                or self._has_reasoning_edge(b, a)
            ):
                continue

            self._add_claim_to_edge(
                a,
                b,
                "co_occurs",
                chunk_id,
                document_id,
                text,
                0.25,
                edge_type="cooccurrence",
            )

    def _has_reasoning_edge(self, u: str, v: str) -> bool:
        if not self.graph.has_edge(u, v):
            return False
        return any(
            data.get("edge_type") == "svo"
            for data in self.graph[u][v].values()
        )

    def _add_or_update_node(
        self,
        node_id: str,
        chunk_id: str,
        document_id: str,
        ent_type: Optional[str] = None,
        mention: Optional[dict] = None,
    ) -> None:
        if node_id not in self.graph:
            self.graph.add_node(
                node_id,
                canonical_name=node_id,
                chunks=set(),
                documents=set(),
                ent_type=ent_type,
                aliases=set(),
                mentions=[],
            )

        data = self.graph.nodes[node_id]
        data.setdefault("chunks", set()).add(chunk_id)
        data.setdefault("documents", set()).add(document_id)

        if ent_type and not data.get("ent_type"):
            data["ent_type"] = ent_type

        if mention:
            data.setdefault("mentions", []).append(mention)
            data.setdefault("aliases", set()).add(mention["text"])

    # ------------------------------------------------------------------
    # Query-time helpers for ECAP-RAG
    # ------------------------------------------------------------------
    def get_claims(self, u: str, v: str, relation: Optional[str] = None):
        """Return source-grounded claims for a graph edge."""
        if not self.graph.has_edge(u, v):
            return []

        claims = []
        for key, data in self.graph[u][v].items():
            if relation and key != relation:
                continue
            claims.extend(data.get("claims", {}).values())
        return claims

    def reasoning_edges(self):
        """Iterate only over SVO/reasoning edges, excluding co-occurrence."""
        for u, v, key, data in self.graph.edges(
            keys=True, data=True
        ):
            if data.get("edge_type") == "svo":
                yield u, v, key, data

    # ------------------------------------------------------------------
    # Persistence
    # ------------------------------------------------------------------
    def save(self, filepath: str = "data/graph_ecap.pkl") -> None:
        os.makedirs(os.path.dirname(filepath) or ".", exist_ok=True)
        with open(filepath, "wb") as f:
            pickle.dump(
                {
                    "graph": self.graph,
                    "alias_index": self._alias_index,
                    "version": "ecap-v3",
                },
                f,
            )

    @classmethod
    def load(
        cls,
        filepath: str = "data/graph_ecap.pkl",
        spacy_model: str = "en_core_web_sm",
    ):
        builder = cls(spacy_model=spacy_model)
        with open(filepath, "rb") as f:
            state = pickle.load(f)

        builder.graph = state["graph"]
        builder._alias_index = state["alias_index"]
        return builder

    def stats(self) -> Dict:
        components = list(nx.weakly_connected_components(self.graph))
        components.sort(key=len, reverse=True)

        reasoning_edge_count = sum(
            1 for _ in self.reasoning_edges()
        )

        claim_count = 0
        for _, _, _, data in self.reasoning_edges():
            claim_count += len(data.get("claims", {}))

        return {
            "num_nodes": self.graph.number_of_nodes(),
            "num_edges": self.graph.number_of_edges(),
            "reasoning_edges": reasoning_edge_count,
            "claims": claim_count,
            "num_components": len(components),
            "main_component_size": len(components[0]) if components else 0,
        }

    def print_fragmentation_report(self) -> None:
        components = list(nx.weakly_connected_components(self.graph))
        components.sort(key=len, reverse=True)

        print(
            f"{len(components)} weakly-connected component(s); "
            f"main component has "
            f"{len(components[0]) if components else 0} node(s)."
        )

        for i, comp in enumerate(components[1:], start=2):
            print(
                f"  Component {i} ({len(comp)} node(s)): "
                f"{sorted(comp)}"
            )


def build_graph_from_documents(
    documents: Dict[str, str],
    output_path: str = "data/graph_ecap.pkl",
    chunk_size_words: int = 250,
) -> KnowledgeGraphBuilder:
    """Build a provenance-aware ECAP-RAG graph from {document_id: text}."""
    builder = KnowledgeGraphBuilder()

    for document_id, text in documents.items():
        for chunk_id, chunk_text in builder.chunk_document(
            text, document_id, chunk_size_words
        ):
            builder.add_chunk(
                chunk_id,
                chunk_text,
                document_id=document_id,
            )

    builder.save(output_path)
    return builder


if __name__ == "__main__":
    import json

    input_path = "ecap_synthetic_documents.json"

    with open(input_path, encoding="utf-8") as f:
        documents = json.load(f)

    builder = build_graph_from_documents(
        documents,
        output_path="data/graph_ecap.pkl",
        chunk_size_words=250,
    )

    print("Stats:", builder.stats())
    builder.print_fragmentation_report()

    print("\nSample reasoning edges:")
    for u, v, key, data in list(builder.reasoning_edges())[:25]:
        print(
            f"{u} --[{key}]--> {v} | "
            f"claims={len(data.get('claims', {}))}"
        )
```

```python
# ============================================================
# ECAP-RAG — HOTPOTQA KNOWLEDGE GRAPH GENERATION
# Single-cell version
# Input: /content/hotpotqa_sample.json
# Output: /content/hotpotqa_graph_clean.pkl
# ============================================================

# -----------------------------
# 1. Imports
# -----------------------------

import os
import re
import json
import pickle
import hashlib
from itertools import combinations

import networkx as nx
import matplotlib.pyplot as plt
import spacy


# -----------------------------
# 2. Configuration
# -----------------------------

FILE_PATH = "/content/hotpotqa_sample.json"
OUTPUT_PATH = "/content/hotpotqa_graph_clean.pkl"

CHUNK_SIZE_WORDS = 250

# -----------------------------
# 3. Load spaCy
# -----------------------------

try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Installing spaCy English model...")
    os.system("python -m spacy download en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")


# ============================================================
# 4. KNOWLEDGE GRAPH BUILDER
# ============================================================

class KnowledgeGraphBuilder:

    TARGET_ENT_TYPES = {
        "PERSON",
        "ORG",
        "WORK_OF_ART",
        "DATE",
        "GPE",
        "LOC",
        "NORP",
        "EVENT",
        "FAC",
    }

    SVO_EDGE_WEIGHT = 1.0
    COOCCUR_EDGE_WEIGHT = 0.25

    # Obvious parser artifacts
    INVALID_NODE_TEXT = {
        "it", "he", "she", "they", "them",
        "their", "his", "her", "its",
        "we", "us", "our",
        "you", "your",
        "i", "me", "my",
        "this", "that", "these", "those",
        "which", "who", "whom", "what",
        "where", "when", "there", "here",
        "someone", "something"
    }

    # Generic noun phrases that usually create noisy nodes
    GENERIC_NODE_TEXT = {
        "the format",
        "the middle child",
        "the race",
        "the tale",
        "the home",
        "the number",
        "the relationship",
        "the odor",
        "the work",
        "the story",
        "the show",
        "the series",
        "the film",
        "the article",
        "the book",
        "the company",
        "the team",
        "a hotel",
        "a genus",
        "a partnership",
        "a chemical compound",
        "a fictional character",
        "some authors",
        "most",
        "part",
        "work"
    }

    def __init__(self):

        # MultiDiGraph allows multiple relation types
        # between the same pair of nodes.
        self.graph = nx.MultiDiGraph()

        # normalized_name -> canonical display name
        self._alias_index = {}

    # ========================================================
    # NORMALIZATION
    # ========================================================

    def _canonical_key(self, text):

        cleaned = text.strip().lower()

        cleaned = re.sub(
            r"^(the|a|an)\s+",
            "",
            cleaned
        )

        cleaned = re.sub(
            r"['\u2019]s\b",
            "",
            cleaned
        )

        cleaned = re.sub(
            r"[^\w\s.\-]",
            "",
            cleaned
        )

        cleaned = re.sub(
            r"\s+",
            " ",
            cleaned
        ).strip()

        return cleaned

    def _resolve_node_id(self, text):

        key = self._canonical_key(text)

        if not key:
            key = text.strip().lower()

        if key not in self._alias_index:
            self._alias_index[key] = text.strip()

        return self._alias_index[key]

    # ========================================================
    # NODE CLEANING
    # ========================================================

    def _is_valid_graph_node(self, text):

        cleaned = re.sub(
            r"\s+",
            " ",
            text.strip()
        )

        key = cleaned.lower()

        if not cleaned:
            return False

        if len(cleaned) < 2:
            return False

        if key in self.INVALID_NODE_TEXT:
            return False

        if key in self.GENERIC_NODE_TEXT:
            return False

        # Remove bare pronouns
        if key in {
            "he", "she", "it", "they",
            "them", "this", "that",
            "these", "those", "which",
            "who", "whom"
        }:
            return False

        # Remove one-word lowercase parser fragments.
        #
        # Keep proper names such as:
        # "London", "Beethoven", "Simpsons"
        if len(cleaned.split()) == 1 and cleaned.islower():
            return False

        return True

    # ========================================================
    # CHUNKING
    # ========================================================

    def chunk_document(
        self,
        text,
        document_id,
        chunk_size_words=250
    ):

        doc = nlp(text)

        chunks = []

        current_sentences = []
        current_words = 0
        chunk_number = 0

        for sent in doc.sents:

            sentence = sent.text.strip()

            if not sentence:
                continue

            word_count = len(sentence.split())

            if (
                current_sentences
                and current_words + word_count > chunk_size_words
            ):

                chunk_number += 1

                chunks.append(
                    (
                        f"{document_id}_chunk_{chunk_number}",
                        " ".join(current_sentences)
                    )
                )

                current_sentences = []
                current_words = 0

            current_sentences.append(sentence)
            current_words += word_count

        if current_sentences:

            chunk_number += 1

            chunks.append(
                (
                    f"{document_id}_chunk_{chunk_number}",
                    " ".join(current_sentences)
                )
            )

        return chunks

    # ========================================================
    # ENTITY EXTRACTION
    # ========================================================

    def _extract_entities(self, doc):

        entities = []

        for ent in doc.ents:

            if ent.label_ in self.TARGET_ENT_TYPES:

                entities.append(
                    (
                        ent.text,
                        ent.label_,
                        ent.start_char,
                        ent.end_char
                    )
                )

        return entities

    # ========================================================
    # NOUN PHRASE / TOKEN SPAN
    # ========================================================

    def _span_for_token(self, token):

        for chunk in token.doc.noun_chunks:

            if chunk.start <= token.i < chunk.end:
                return chunk.text

        tokens = sorted(
            [
                t
                for t in token.subtree
                if (
                    t.dep_ in {
                        "compound",
                        "amod",
                        "poss"
                    }
                    or t.i == token.i
                )
            ],
            key=lambda x: x.i
        )

        if tokens:
            return " ".join(t.text for t in tokens)

        return token.text

    # ========================================================
    # CONJUNCTION
    # ========================================================

    def _expand_conjuncts(self, token):

        result = [token]

        for child in token.children:

            if (
                child.dep_ == "conj"
                and child.pos_ == token.pos_
            ):
                result.append(child)

        return result

    # ========================================================
    # PASSIVE AGENT
    # ========================================================

    def _find_agent(self, verb):

        for child in verb.children:

            if child.dep_ == "agent":

                pobj = next(
                    (
                        gc
                        for gc in child.children
                        if gc.dep_ == "pobj"
                    ),
                    None
                )

                if pobj is not None:
                    return pobj

        return None

    # ========================================================
    # COMPLEMENT
    # ========================================================

    def _find_complement(
        self,
        token,
        dependency_types
    ):

        return next(
            (
                child
                for child in token.children
                if child.dep_ in dependency_types
            ),
            None
        )

    # ========================================================
    # SVO EXTRACTION
    # ========================================================

    def _extract_svo_triples(self, doc):

        triples = []

        for sent in doc.sents:

            sentence_text = sent.text.strip()

            for token in sent:

                # ------------------------------------------------
                # COPULAR
                # ------------------------------------------------

                is_copula = (
                    token.pos_ == "AUX"
                    and token.dep_ == "ROOT"
                )

                if not is_copula and token.pos_ != "VERB":
                    continue

                if is_copula:

                    subject = self._find_complement(
                        token,
                        {"nsubj", "nsubjpass"}
                    )

                    complement = self._find_complement(
                        token,
                        {"attr", "acomp"}
                    )

                    if (
                        subject is None
                        or complement is None
                    ):
                        continue

                    for s in self._expand_conjuncts(subject):

                        for o in self._expand_conjuncts(
                            complement
                        ):

                            triples.append(
                                (
                                    self._span_for_token(s),
                                    "is",
                                    self._span_for_token(o),
                                    sentence_text,
                                    1.0
                                )
                            )

                    continue

                # ------------------------------------------------
                # SUBJECT
                # ------------------------------------------------

                subject = self._find_complement(
                    token,
                    {"nsubj", "nsubjpass"}
                )

                if subject is None:
                    continue

                # ------------------------------------------------
                # PASSIVE
                # ------------------------------------------------

                is_passive = any(
                    child.dep_ == "nsubjpass"
                    for child in token.children
                )

                if is_passive:

                    agent = self._find_agent(token)

                    relation = token.lemma_.lower()

                    if agent is not None:

                        for obj in self._expand_conjuncts(
                            subject
                        ):

                            triples.append(
                                (
                                    self._span_for_token(agent),
                                    relation,
                                    self._span_for_token(obj),
                                    sentence_text,
                                    1.0
                                )
                            )

                        continue

                    continue

                # ------------------------------------------------
                # ACTIVE OBJECT
                # ------------------------------------------------

                obj = self._find_complement(
                    token,
                    {
                        "dobj",
                        "attr",
                        "oprd"
                    }
                )

                # ------------------------------------------------
                # PREPOSITIONAL OBJECT
                # ------------------------------------------------

                if obj is None:

                    for child in token.children:

                        if child.dep_ == "prep":

                            pobj = next(
                                (
                                    gc
                                    for gc in child.children
                                    if gc.dep_ == "pobj"
                                ),
                                None
                            )

                            if pobj is not None:

                                obj = pobj
                                break

                if obj is None:
                    continue

                relation = token.lemma_.lower()

                score = (
                    1.0
                    if obj.dep_ in {
                        "dobj",
                        "attr",
                        "oprd"
                    }
                    else 0.85
                )

                for s in self._expand_conjuncts(subject):

                    for o in self._expand_conjuncts(obj):

                        triples.append(
                            (
                                self._span_for_token(s),
                                relation,
                                self._span_for_token(o),
                                sentence_text,
                                score
                            )
                        )

        return triples

    # ========================================================
    # CLAIM ID
    # ========================================================

    def _make_claim_id(
        self,
        document_id,
        chunk_id,
        sentence,
        subject,
        relation,
        object_
    ):

        raw = "||".join(
            [
                document_id,
                chunk_id,
                sentence,
                subject,
                relation,
                object_
            ]
        )

        digest = hashlib.sha1(
            raw.encode("utf-8")
        ).hexdigest()[:16]

        return f"claim_{digest}"

    # ========================================================
    # ADD CLAIM
    # ========================================================

    def _add_claim_to_edge(
        self,
        subject,
        object_,
        relation,
        chunk_id,
        document_id,
        source_span,
        extraction_score,
        edge_type="svo"
    ):

        claim_id = self._make_claim_id(
            document_id,
            chunk_id,
            source_span,
            subject,
            relation,
            object_
        )

        claim = {

            "claim_id": claim_id,

            "document_id": document_id,

            "chunk_id": chunk_id,

            "source_span": source_span,

            "subject": subject,

            "relation": relation,

            "object": object_,

            "extraction_score": extraction_score,

            # Reserved for next ECAP-RAG stage
            "polarity": "positive"
        }

        edge_key = relation

        if self.graph.has_edge(
            subject,
            object_,
            key=edge_key
        ):

            data = self.graph[
                subject
            ][
                object_
            ][
                edge_key
            ]

            data.setdefault(
                "claims",
                {}
            )

            data["claims"][
                claim_id
            ] = claim

            data.setdefault(
                "chunks",
                set()
            ).add(chunk_id)

            data["weight"] = (
                data.get("weight", 0)
                + (
                    self.SVO_EDGE_WEIGHT
                    if edge_type == "svo"
                    else self.COOCCUR_EDGE_WEIGHT
                )
            )

        else:

            self.graph.add_edge(
                subject,
                object_,
                key=edge_key,

                relation=relation,

                edge_type=edge_type,

                chunks={chunk_id},

                weight=(
                    self.SVO_EDGE_WEIGHT
                    if edge_type == "svo"
                    else self.COOCCUR_EDGE_WEIGHT
                ),

                claims={
                    claim_id: claim
                }
            )

    # ========================================================
    # ADD CHUNK
    # ========================================================

    def add_chunk(
        self,
        chunk_id,
        text,
        document_id=None
    ):

        if document_id is None:

            document_id = (
                chunk_id.split(
                    "_chunk_"
                )[0]
            )

        doc = nlp(text)

        entities = self._extract_entities(doc)

        # ----------------------------------------------------
        # SVO RELATIONSHIPS
        # ----------------------------------------------------

        triples = self._extract_svo_triples(doc)

        for (
            subject_text,
            relation,
            object_text,
            sentence,
            extraction_score
        ) in triples:

            if not self._is_valid_graph_node(
                subject_text
            ):
                continue

            if not self._is_valid_graph_node(
                object_text
            ):
                continue

            subject = self._resolve_node_id(
                subject_text
            )

            object_ = self._resolve_node_id(
                object_text
            )

            if subject == object_:
                continue

            self._add_or_update_node(
                subject,
                chunk_id,
                document_id
            )

            self._add_or_update_node(
                object_,
                chunk_id,
                document_id
            )

            self._add_claim_to_edge(
                subject,
                object_,
                relation,
                chunk_id,
                document_id,
                sentence,
                extraction_score,
                "svo"
            )

        # ----------------------------------------------------
        # NAMED ENTITIES
        # ----------------------------------------------------

        valid_entity_nodes = []

        for (
            entity_text,
            entity_type,
            start_char,
            end_char
        ) in entities:

            if not self._is_valid_graph_node(
                entity_text
            ):
                continue

            node_id = self._resolve_node_id(
                entity_text
            )

            self._add_or_update_node(
                node_id,
                chunk_id,
                document_id,
                entity_type,
                {
                    "document_id": document_id,
                    "chunk_id": chunk_id,
                    "text": entity_text,
                    "start_char": start_char,
                    "end_char": end_char
                }
            )

            valid_entity_nodes.append(node_id)

        # ----------------------------------------------------
        # CO-OCCURRENCE
        # ----------------------------------------------------

        valid_entity_nodes = sorted(
            set(valid_entity_nodes)
        )

        for a, b in combinations(
            valid_entity_nodes,
            2
        ):

            # Don't add co-occurrence if a real
            # reasoning edge already exists.
            if (
                self._has_reasoning_edge(a, b)
                or self._has_reasoning_edge(b, a)
            ):
                continue

            self._add_claim_to_edge(
                a,
                b,
                "co_occurs",
                chunk_id,
                document_id,
                text,
                0.25,
                "cooccurrence"
            )

    # ========================================================
    # NODE UPDATE
    # ========================================================

    def _add_or_update_node(
        self,
        node_id,
        chunk_id,
        document_id,
        ent_type=None,
        mention=None
    ):

        if node_id not in self.graph:

            self.graph.add_node(
                node_id,

                canonical_name=node_id,

                chunks=set(),

                documents=set(),

                ent_type=ent_type,

                aliases=set(),

                mentions=[]
            )

        data = self.graph.nodes[
            node_id
        ]

        data.setdefault(
            "chunks",
            set()
        ).add(chunk_id)

        data.setdefault(
            "documents",
            set()
        ).add(document_id)

        if (
            ent_type
            and not data.get("ent_type")
        ):
            data["ent_type"] = ent_type

        if mention:

            data.setdefault(
                "mentions",
                []
            ).append(mention)

            data.setdefault(
                "aliases",
                set()
            ).add(
                mention["text"]
            )

    # ========================================================
    # REASONING EDGES
    # ========================================================

    def _has_reasoning_edge(
        self,
        u,
        v
    ):

        if not self.graph.has_edge(
            u,
            v
        ):
            return False

        return any(
            data.get("edge_type") == "svo"
            for data in self.graph[
                u
            ][
                v
            ].values()
        )

    def reasoning_edges(self):

        for (
            u,
            v,
            key,
            data
        ) in self.graph.edges(
            keys=True,
            data=True
        ):

            if data.get(
                "edge_type"
            ) == "svo":

                yield (
                    u,
                    v,
                    key,
                    data
                )

    # ========================================================
    # BUILD GRAPH
    # ========================================================

    def build_from_documents(
        self,
        documents,
        chunk_size_words=250
    ):

        for document_id, text in documents.items():

            chunks = self.chunk_document(
                text,
                document_id,
                chunk_size_words
            )

            for chunk_id, chunk_text in chunks:

                self.add_chunk(
                    chunk_id,
                    chunk_text,
                    document_id
                )

        return self

    # ========================================================
    # STATISTICS
    # ========================================================

    def stats(self):

        components = list(
            nx.weakly_connected_components(
                self.graph
            )
        )

        components.sort(
            key=len,
            reverse=True
        )

        reasoning_edge_count = sum(
            1
            for _ in self.reasoning_edges()
        )

        claim_count = 0

        for (
            _,
            _,
            _,
            data
        ) in self.reasoning_edges():

            claim_count += len(
                data.get(
                    "claims",
                    {}
                )
            )

        return {

            "num_nodes":
                self.graph.number_of_nodes(),

            "num_edges":
                self.graph.number_of_edges(),

            "reasoning_edges":
                reasoning_edge_count,

            "claims":
                claim_count,

            "num_components":
                len(components),

            "main_component_size":
                (
                    len(components[0])
                    if components
                    else 0
                )
        }

    # ========================================================
    # SAVE
    # ========================================================

    def save(
        self,
        path
    ):

        with open(
            path,
            "wb"
        ) as f:

            pickle.dump(
                {
                    "graph":
                        self.graph,

                    "alias_index":
                        self._alias_index,

                    "version":
                        "ECAP-RAG-clean-v2"
                },
                f
            )


# ============================================================
# 5. LOAD HOTPOTQA
# ============================================================

print("=" * 70)
print("LOADING HOTPOTQA")
print("=" * 70)

if not os.path.exists(FILE_PATH):

    raise FileNotFoundError(
        f"Dataset not found: {FILE_PATH}"
    )

with open(
    FILE_PATH,
    "r",
    encoding="utf-8"
) as f:

    hotpotqa = json.load(f)

print(
    "Questions:",
    len(hotpotqa)
)


# ============================================================
# 6. CONVERT HOTPOTQA → DOCUMENTS
# ============================================================

documents = {}

for item in hotpotqa:

    question_id = item["id"]

    titles = item["context"]["title"]

    sentence_groups = item["context"]["sentences"]

    for title, sentences in zip(
        titles,
        sentence_groups
    ):

        document_id = (
            f"{question_id}__{title}"
        )

        documents[
            document_id
        ] = " ".join(sentences)


print(
    "Documents:",
    len(documents)
)


# ============================================================
# 7. BUILD KG
# ============================================================

print("\n")
print("=" * 70)
print("BUILDING ECAP-RAG KNOWLEDGE GRAPH")
print("=" * 70)

builder = KnowledgeGraphBuilder()

builder.build_from_documents(
    documents,
    chunk_size_words=CHUNK_SIZE_WORDS
)

builder.save(
    OUTPUT_PATH
)


# ============================================================
# 8. PRINT STATISTICS
# ============================================================

print("\n")
print("=" * 70)
print("GRAPH STATISTICS")
print("=" * 70)

stats = builder.stats()

for key, value in stats.items():

    print(
        f"{key:25s}: {value}"
    )


# ============================================================
# 9. PRINT REASONING EDGES + CLAIMS
# ============================================================

print("\n")
print("=" * 70)
print("RELATIONSHIPS AND CLAIMS")
print("=" * 70)

for (
    u,
    v,
    relation,
    data
) in builder.reasoning_edges():

    claims = data.get(
        "claims",
        {}
    )

    print("\n")
    print(
        f"{u} --[{relation}]--> {v}"
    )

    print(
        f"Claims: {len(claims)}"
    )

    for claim in claims.values():

        print(
            f"  [{claim['document_id']}]"
        )

        print(
            f"  Evidence: "
            f"{claim['source_span']}"
        )


# ============================================================
# 10. CREATE VISUALIZATION
# ============================================================

print("\n")
print("=" * 70)
print("GENERATING GRAPH VISUALIZATION")
print("=" * 70)

G = nx.DiGraph()

for (
    u,
    v,
    relation,
    data
) in builder.reasoning_edges():

    G.add_edge(
        u,
        v,
        label=relation
    )


# ------------------------------------------------------------
# Limit visualization to largest connected component
# ------------------------------------------------------------

if len(G) > 0:

    components = sorted(
        nx.weakly_connected_components(G),
        key=len,
        reverse=True
    )

    largest_component = components[0]

    G_visual = G.subgraph(
        largest_component
    ).copy()

else:

    G_visual = G


# ------------------------------------------------------------
# Draw
# ------------------------------------------------------------

plt.figure(
    figsize=(22, 16)
)

pos = nx.spring_layout(
    G_visual,
    seed=42,
    k=2.0
)

nx.draw_networkx_nodes(
    G_visual,
    pos,
    node_size=1800
)

nx.draw_networkx_edges(
    G_visual,
    pos,
    arrows=True,
    arrowsize=15,
    width=1.2
)

nx.draw_networkx_labels(
    G_visual,
    pos,
    font_size=8
)

edge_labels = nx.get_edge_attributes(
    G_visual,
    "label"
)

nx.draw_networkx_edge_labels(
    G_visual,
    pos,
    edge_labels=edge_labels,
    font_size=7
)

plt.title(
    "ECAP-RAG — Clean HotpotQA Knowledge Graph",
    fontsize=16
)

plt.axis("off")

plt.tight_layout()

plt.show()


# ============================================================
# 11. FINISHED
# ============================================================

print("\n")
print("=" * 70)
print("DONE")
print("=" * 70)

print(
    f"KG saved to: {OUTPUT_PATH}"
)

print(
    "Next stage: question-specific subgraphs + path retrieval."
)
```

```python
# ============================================================
# FULL ECAP-RAG KNOWLEDGE GRAPH VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt
import networkx as nx

print("=" * 70)
print("GENERATING FULL ECAP-RAG KNOWLEDGE GRAPH")
print("=" * 70)

# ------------------------------------------------------------
# 1. Create visualization graph from ALL reasoning edges
# ------------------------------------------------------------

full_graph = nx.Graph()

for u, v, key, data in builder.graph.edges(
    keys=True,
    data=True
):

    # Add nodes even if they have unusual connectivity
    full_graph.add_node(u)
    full_graph.add_node(v)

    # Graph visualization uses an undirected edge
    # while the actual ECAP-RAG graph remains directed.
    full_graph.add_edge(
        u,
        v,
        relation=data.get("relation", "")
    )

print(f"Nodes plotted : {full_graph.number_of_nodes()}")
print(f"Edges plotted : {full_graph.number_of_edges()}")

# ------------------------------------------------------------
# 2. Layout
# ------------------------------------------------------------

print("Calculating layout...")

pos = nx.spring_layout(
    full_graph,
    seed=42,
    k=0.35,
    iterations=200
)

# ------------------------------------------------------------
# 3. Figure
# ------------------------------------------------------------

plt.figure(figsize=(32, 24))

# ------------------------------------------------------------
# 4. Draw ALL nodes
# ------------------------------------------------------------

nx.draw_networkx_nodes(
    full_graph,
    pos,
    node_size=80,
    alpha=0.85
)

# ------------------------------------------------------------
# 5. Draw ALL edges
# ------------------------------------------------------------

nx.draw_networkx_edges(
    full_graph,
    pos,
    width=0.5,
    alpha=0.30
)

# ------------------------------------------------------------
# 6. Labels
# ------------------------------------------------------------
# With 1,000+ nodes, labeling every node makes the figure
# unreadable. Set SHOW_ALL_LABELS=True if you really want it.

SHOW_ALL_LABELS = False

if SHOW_ALL_LABELS:

    nx.draw_networkx_labels(
        full_graph,
        pos,
        font_size=4
    )

# ------------------------------------------------------------
# 7. Edge relation labels
# ------------------------------------------------------------
# Also disabled by default because there are many edges.

SHOW_EDGE_LABELS = False

if SHOW_EDGE_LABELS:

    edge_labels = nx.get_edge_attributes(
        full_graph,
        "relation"
    )

    nx.draw_networkx_edge_labels(
        full_graph,
        pos,
        edge_labels=edge_labels,
        font_size=3
    )

# ------------------------------------------------------------
# 8. Title
# ------------------------------------------------------------

plt.title(
    "ECAP-RAG — Full HotpotQA Knowledge Graph\n"
    f"{full_graph.number_of_nodes()} nodes / "
    f"{full_graph.number_of_edges()} reasoning edges",
    fontsize=20
)

plt.axis("off")
plt.tight_layout()

# ------------------------------------------------------------
# 9. Save full graph image
# ------------------------------------------------------------

output_path = "/content/ecap_rag_full_graph.png"

plt.savefig(
    output_path,
    dpi=250,
    bbox_inches="tight"
)

plt.show()

print()
print("=" * 70)
print("FULL GRAPH GENERATED")
print("=" * 70)
print(f"Saved to: {output_path}")
```

```python
# ============================================================
# ECAP-RAG — CONNECTED ENTITY-ANCHORED KNOWLEDGE GRAPH
# ============================================================
#
# Main improvement over previous version:
#
#   Sentence
#       ↓
#   Named entities
#       ↓
#   Dependency relation
#       ↓
#   Entity-to-entity edge
#
# We do NOT create arbitrary noun phrases as graph nodes
# unless they are useful entity/value nodes.
#
# This is designed for:
#
#   KG -> Path Retrieval -> Evidence Retrieval
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import re
import json
import pickle
import hashlib
from collections import defaultdict, Counter

import networkx as nx
import matplotlib.pyplot as plt

try:
    import spacy
except ImportError:
    !pip -q install spacy
    import spacy


# ============================================================
# 2. LOAD SPACY
# ============================================================

MODEL_NAME = "en_core_web_sm"

try:
    nlp = spacy.load(MODEL_NAME)
except OSError:
    !python -m spacy download en_core_web_sm
    nlp = spacy.load(MODEL_NAME)

print("spaCy:", spacy.__version__)
print("Model:", MODEL_NAME)


# ============================================================
# 3. CONFIG
# ============================================================

DATA_PATH = "/content/hotpotqa_sample.json"

KG_PATH = "/content/ecap_rag_connected_kg.pkl"

PLOT_PATH = (
    "/content/ecap_rag_connected_graph.png"
)


# ============================================================
# 4. VALID ENTITY TYPES
# ============================================================

ENTITY_TYPES = {
    "PERSON",
    "ORG",
    "GPE",
    "LOC",
    "FAC",
    "EVENT",
    "WORK_OF_ART",
    "NORP",
    "PRODUCT",
    "LAW",
    "LANGUAGE",
    "DATE",
    "TIME",
    "MONEY",
    "QUANTITY",
    "CARDINAL",
    "ORDINAL",
}


# ============================================================
# 5. BAD ENTITY VALUES
# ============================================================

BAD_ENTITIES = {
    "it",
    "he",
    "she",
    "they",
    "them",
    "their",
    "his",
    "her",
    "its",
    "we",
    "us",
    "our",
    "you",
    "your",
    "i",
    "me",
    "my",
    "this",
    "that",
    "these",
    "those",
    "which",
    "who",
    "whom",
    "what",
    "where",
    "when",
    "there",
    "here",
}


# ============================================================
# 6. TEXT NORMALIZATION
# ============================================================

def clean_text(text):

    if text is None:
        return ""

    text = str(text)

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def normalize_entity(text):

    text = clean_text(text)

    text = text.strip(
        " \t\n\r.,;:!?()[]{}\"'"
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def valid_entity(text):

    text = normalize_entity(text)

    if not text:
        return False

    low = text.lower()

    if low in BAD_ENTITIES:
        return False

    if "generator object" in low:
        return False

    if "_cython_" in low:
        return False

    if len(text) < 2:
        return False

    if not re.search(
        r"[A-Za-z0-9]",
        text
    ):
        return False

    return True


# ============================================================
# 7. RELATION NORMALIZATION
# ============================================================

def normalize_relation(relation):

    relation = clean_text(
        relation
    ).lower()

    relation = relation.replace(
        " ",
        "_"
    )

    replacements = {

        "was": "be",
        "were": "be",
        "is": "be",
        "are": "be",

        "has": "have",
        "had": "have",

        "started": "start",
        "starts": "start",

        "ended": "end",
        "ends": "end",

        "founded": "found",
        "founds": "found",

        "formed": "form",
        "forms": "form",

        "created": "create",
        "creates": "create",

        "joined": "join",
        "joins": "join",

        "played": "play",
        "plays": "play",

        "released": "release",
        "releases": "release",

        "published": "publish",
        "publishes": "publish",

        "directed": "direct",
        "directs": "direct",

        "written": "write",
        "wrote": "write",

        "won": "win",
        "wins": "win",

        "lost": "lose",
        "loses": "lose",

        "contains": "contain",
        "contained": "contain",

        "includes": "include",
        "included": "include",

        "features": "feature",
        "featured": "feature",

        "produced": "produce",
        "produces": "produce",

        "announced": "announce",
        "announces": "announce",

        "acquired": "acquire",
        "acquires": "acquire",

        "defeated": "defeat",
        "defeats": "defeat",

        "reached": "reach",
        "reaches": "reach",
    }

    return replacements.get(
        relation,
        relation
    )


# ============================================================
# 8. CLAIM ID
# ============================================================

def make_claim_id(
    document_id,
    sentence_id,
    subject,
    relation,
    obj,
):

    raw = (
        f"{document_id}|"
        f"{sentence_id}|"
        f"{subject}|"
        f"{relation}|"
        f"{obj}"
    )

    return hashlib.sha1(
        raw.encode(
            "utf-8"
        )
    ).hexdigest()[:16]


# ============================================================
# 9. KNOWLEDGE GRAPH
# ============================================================

class ConnectedECAPRAGKG:

    def __init__(self, nlp):

        self.nlp = nlp

        self.graph = nx.MultiDiGraph()

        self.claims = {}

        self.question_documents = (
            defaultdict(list)
        )

        self.entity_mentions = (
            defaultdict(int)
        )


    # ========================================================
    # ENTITY
    # ========================================================

    def add_entity(
        self,
        text,
        entity_type,
        question_id,
        document_id,
    ):

        text = normalize_entity(
            text
        )

        if not valid_entity(
            text
        ):
            return None

        if text not in self.graph:

            self.graph.add_node(

                text,

                entity_type=entity_type,

                entity_types=set(),

                questions=set(),

                documents=set(),

                mentions=0,
            )

        node = self.graph.nodes[
            text
        ]

        node[
            "mentions"
        ] += 1

        node[
            "entity_types"
        ].add(
            entity_type
        )

        node[
            "questions"
        ].add(
            question_id
        )

        node[
            "documents"
        ].add(
            document_id
        )

        self.entity_mentions[
            text
        ] += 1

        return text


    # ========================================================
    # EXTRACT ENTITIES
    # ========================================================

    def get_sentence_entities(
        self,
        doc,
        question_id,
        document_id,
    ):

        entities = []

        for ent in doc.ents:

            text = normalize_entity(
                ent.text
            )

            if not valid_entity(
                text
            ):
                continue

            entity = self.add_entity(

                text,

                ent.label_,

                question_id,

                document_id,
            )

            if entity:

                entities.append(
                    (
                        entity,
                        ent.start,
                        ent.end,
                        ent.label_
                    )
                )

        return entities


    # ========================================================
    # FIND ENTITY INSIDE SPAN
    # ========================================================

    def entity_inside_span(
        self,
        start,
        end,
        entities,
    ):

        candidates = []

        for (
            entity,
            ent_start,
            ent_end,
            ent_type
        ) in entities:

            if (
                ent_start >= start
                and
                ent_end <= end
            ):

                candidates.append(
                    (
                        entity,
                        ent_start,
                        ent_end,
                        ent_type
                    )
                )

        if not candidates:
            return None

        # Prefer the longest entity span
        candidates.sort(
            key=lambda x:
                x[2] - x[1],
            reverse=True
        )

        return candidates[0][0]


    # ========================================================
    # FIND SUBJECT ENTITY
    # ========================================================

    def find_subject_entity(
        self,
        verb,
        doc,
        entities,
        previous_entities,
    ):

        # --------------------------------------------
        # Direct grammatical subject
        # --------------------------------------------

        subject_tokens = []

        for child in verb.children:

            if child.dep_ in {
                "nsubj",
                "nsubjpass",
                "csubj",
            }:

                subject_tokens = list(
                    child.subtree
                )

                break

        if subject_tokens:

            start = min(
                t.i
                for t in subject_tokens
            )

            end = max(
                t.i
                for t in subject_tokens
            ) + 1

            entity = (
                self.entity_inside_span(
                    start,
                    end,
                    entities
                )
            )

            if entity:
                return entity

            # Pronoun fallback
            text = " ".join(
                t.text
                for t in subject_tokens
            ).lower()

            if (
                text in BAD_ENTITIES
                and previous_entities
            ):

                return previous_entities[-1]

        # --------------------------------------------
        # Passive voice fallback
        # --------------------------------------------

        for child in verb.children:

            if child.dep_ == "nsubjpass":

                tokens = list(
                    child.subtree
                )

                if tokens:

                    start = min(
                        t.i
                        for t in tokens
                    )

                    end = max(
                        t.i
                        for t in tokens
                    ) + 1

                    entity = (
                        self.entity_inside_span(
                            start,
                            end,
                            entities
                        )
                    )

                    if entity:
                        return entity

        return None


    # ========================================================
    # FIND OBJECT ENTITIES
    # ========================================================

    def find_object_entities(
        self,
        verb,
        doc,
        entities,
    ):

        results = []

        # --------------------------------------------
        # Direct object / attribute
        # --------------------------------------------

        for child in verb.children:

            if child.dep_ in {
                "dobj",
                "obj",
                "attr",
                "oprd",
            }:

                tokens = list(
                    child.subtree
                )

                if not tokens:
                    continue

                start = min(
                    t.i
                    for t in tokens
                )

                end = max(
                    t.i
                    for t in tokens
                ) + 1

                entity = (
                    self.entity_inside_span(
                        start,
                        end,
                        entities
                    )
                )

                if entity:
                    results.append(
                        entity
                    )

        # --------------------------------------------
        # Prepositional object
        # --------------------------------------------

        for child in verb.children:

            if child.dep_ == "prep":

                for pobj in child.children:

                    if pobj.dep_ == "pobj":

                        tokens = list(
                            pobj.subtree
                        )

                        if not tokens:
                            continue

                        start = min(
                            t.i
                            for t in tokens
                        )

                        end = max(
                            t.i
                            for t in tokens
                        ) + 1

                        entity = (
                            self.entity_inside_span(
                                start,
                                end,
                                entities
                            )
                        )

                        if entity:

                            results.append(
                                entity
                            )

        return list(
            dict.fromkeys(
                results
            )
        )


    # ========================================================
    # ENTITY-TO-ENTITY RELATION EXTRACTION
    # ========================================================

    def extract_relations(
        self,
        doc,
        entities,
        question_id,
        document_id,
        sentence_id,
        source_sentence,
        previous_entities,
    ):

        sentence_claims = []

        for token in doc:

            if token.pos_ not in {
                "VERB",
                "AUX",
            }:

                continue

            subject = (
                self.find_subject_entity(
                    token,
                    doc,
                    entities,
                    previous_entities,
                )
            )

            if not subject:
                continue

            objects = (
                self.find_object_entities(
                    token,
                    doc,
                    entities,
                )
            )

            if not objects:
                continue

            relation = normalize_relation(
                token.lemma_
            )

            if not relation:
                continue

            for obj in objects:

                if subject == obj:
                    continue

                claim_id = make_claim_id(

                    document_id,

                    sentence_id,

                    subject,

                    relation,

                    obj,
                )

                claim = {

                    "claim_id":
                        claim_id,

                    "question_id":
                        question_id,

                    "document_id":
                        document_id,

                    "sentence_id":
                        sentence_id,

                    "source_span":
                        source_sentence,

                    "subject":
                        subject,

                    "relation":
                        relation,

                    "object":
                        obj,

                    "polarity":
                        "positive",

                    "extraction_score":
                        1.0,

                    "contradiction_group":
                        None,
                }

                # ------------------------------------
                # Add edge
                # ------------------------------------

                existing_key = None

                if self.graph.has_edge(
                    subject,
                    obj
                ):

                    for key, data in (
                        self.graph[
                            subject
                        ][obj].items()
                    ):

                        if (
                            data.get(
                                "relation"
                            )
                            == relation
                        ):

                            existing_key = key
                            break

                if existing_key is None:

                    self.graph.add_edge(

                        subject,

                        obj,

                        relation=relation,

                        edge_type="reasoning",

                        claims={
                            claim_id:
                                claim
                        },

                        weight=1.0,
                    )

                else:

                    edge = self.graph[
                        subject
                    ][obj][
                        existing_key
                    ]

                    edge[
                        "claims"
                    ][
                        claim_id
                    ] = claim

                    edge[
                        "weight"
                    ] += 1.0

                self.claims[
                    claim_id
                ] = claim

                sentence_claims.append(
                    claim
                )

        return sentence_claims


    # ========================================================
    # ADD DOCUMENT
    # ========================================================

    def add_document(
        self,
        question_id,
        document_id,
        title,
        sentences,
    ):

        self.question_documents[
            question_id
        ].append(
            document_id
        )

        previous_entities = []

        for sentence_id, sentence in enumerate(
            sentences
        ):

            sentence = clean_text(
                sentence
            )

            if not sentence:
                continue

            doc = self.nlp(
                sentence
            )

            entities = (
                self.get_sentence_entities(
                    doc,
                    question_id,
                    document_id,
                )
            )

            current_entities = [
                x[0]
                for x in entities
            ]

            self.extract_relations(

                doc,

                entities,

                question_id,

                document_id,

                sentence_id,

                sentence,

                previous_entities,
            )

            # Keep recent real entities
            previous_entities.extend(
                current_entities
            )

            previous_entities = list(
                dict.fromkeys(
                    previous_entities
                )
            )[-10:]


    # ========================================================
    # BUILD HOTPOTQA
    # ========================================================

    def build_from_hotpotqa(
        self,
        data,
    ):

        for item in data:

            question_id = item[
                "id"
            ]

            context = item[
                "context"
            ]

            for (
                title,
                sentences
            ) in zip(
                context["title"],
                context["sentences"]
            ):

                document_id = (
                    f"{question_id}"
                    f"__{title}"
                )

                self.add_document(

                    question_id,

                    document_id,

                    title,

                    sentences,
                )


    # ========================================================
    # REASONING EDGES
    # ========================================================

    def reasoning_edges(self):

        for (
            u,
            v,
            key,
            data
        ) in self.graph.edges(
            keys=True,
            data=True
        ):

            if (
                data.get(
                    "edge_type"
                )
                == "reasoning"
            ):

                yield (
                    u,
                    v,
                    key,
                    data
                )


    # ========================================================
    # STATS
    # ========================================================

    def stats(self):

        reasoning = list(
            self.reasoning_edges()
        )

        components = list(
            nx.weakly_connected_components(
                self.graph
            )
        )

        components.sort(
            key=len,
            reverse=True
        )

        return {

            "nodes":
                self.graph.number_of_nodes(),

            "edges":
                self.graph.number_of_edges(),

            "reasoning_edges":
                len(reasoning),

            "claims":
                len(self.claims),

            "components":
                len(components),

            "largest_component":
                (
                    len(components[0])
                    if components
                    else 0
                ),

            "entities_with_degree_2plus":
                sum(
                    1
                    for n, d
                    in self.graph.degree()
                    if d >= 2
                ),
        }


    # ========================================================
    # SAVE
    # ========================================================

    def save(
        self,
        path,
    ):

        with open(
            path,
            "wb"
        ) as f:

            pickle.dump(
                self,
                f
            )


# ============================================================
# 11. LOAD DATA
# ============================================================

print()
print("=" * 70)
print("LOADING HOTPOTQA")
print("=" * 70)

with open(
    DATA_PATH,
    "r",
    encoding="utf-8"
) as f:

    hotpotqa = json.load(
        f
    )

print(
    "Questions:",
    len(hotpotqa)
)

print(
    "Documents:",
    sum(
        len(
            item["context"]["title"]
        )
        for item in hotpotqa
    )
)


# ============================================================
# 12. BUILD KG
# ============================================================

print()
print("=" * 70)
print("BUILDING ENTITY-ANCHORED KG")
print("=" * 70)

builder = ConnectedECAPRAGKG(
    nlp
)

builder.build_from_hotpotqa(
    hotpotqa
)


# ============================================================
# 13. STATS
# ============================================================

stats = builder.stats()

print()
print("=" * 70)
print("NEW KG STATISTICS")
print("=" * 70)

for key, value in stats.items():

    print(
        f"{key:<30}: {value}"
    )


# ============================================================
# 14. SHOW SAMPLE CLAIMS
# ============================================================

print()
print("=" * 70)
print("SAMPLE ENTITY-TO-ENTITY CLAIMS")
print("=" * 70)

for claim in list(
    builder.claims.values()
)[:30]:

    print(
        f"{claim['subject']}"
        f" --[{claim['relation']}]--> "
        f"{claim['object']}"
    )

    print(
        "  evidence:",
        claim["source_span"]
    )

    print()


# ============================================================
# 15. CONNECTIVITY ANALYSIS
# ============================================================

G = nx.Graph()

for (
    u,
    v,
    key,
    data
) in builder.reasoning_edges():

    relation = data.get(
        "relation",
        ""
    )

    if G.has_edge(
        u,
        v
    ):

        G[u][v][
            "relations"
        ].add(
            relation
        )

    else:

        G.add_edge(

            u,

            v,

            relations={
                relation
            }
        )


components = sorted(
    nx.connected_components(G),
    key=len,
    reverse=True
)

print()
print("=" * 70)
print("CONNECTIVITY")
print("=" * 70)

print(
    "Nodes:",
    G.number_of_nodes()
)

print(
    "Edges:",
    G.number_of_edges()
)

print(
    "Components:",
    len(components)
)

print(
    "Largest components:",
    [
        len(c)
        for c in components[:20]
    ]
)


# ============================================================
# 16. SAVE
# ============================================================

builder.save(
    KG_PATH
)

print()
print(
    "KG saved:",
    KG_PATH
)


# ============================================================
# 17. PRESENTATION GRAPH
# ============================================================
#
# We now select the largest genuinely connected component.
#
# This is deliberately different from the previous approach.
#
# If the largest component is small, that is a signal that
# the extractor still needs work.
#
# ============================================================

if components:

    largest_component = components[0]

    P = G.subgraph(
        largest_component
    ).copy()

else:

    P = G.copy()


print()
print("=" * 70)
print("PRESENTATION GRAPH")
print("=" * 70)

print(
    "Nodes:",
    P.number_of_nodes()
)

print(
    "Edges:",
    P.number_of_edges()
)


# ============================================================
# 18. VISUALIZE
# ============================================================

if P.number_of_nodes() > 1:

    # Don't try to display thousands of nodes.
    # But don't artificially combine disconnected components.

    if P.number_of_nodes() > 120:

        degree = dict(
            P.degree()
        )

        important = sorted(
            P.nodes(),
            key=lambda x:
                degree[x],
            reverse=True
        )[:120]

        V = P.subgraph(
            important
        ).copy()

    else:

        V = P.copy()


    pos = nx.spring_layout(

        V,

        seed=42,

        k=1.0,

        iterations=300
    )


    degree = dict(
        V.degree()
    )


    node_sizes = [

        150
        +
        degree[n] * 40

        for n in V.nodes()

    ]


    plt.figure(
        figsize=(22, 18)
    )


    nx.draw_networkx_edges(

        V,

        pos,

        width=1.0,

        alpha=0.35
    )


    nx.draw_networkx_nodes(

        V,

        pos,

        node_size=node_sizes,

        alpha=0.9
    )


    # Label most connected nodes
    label_nodes = sorted(

        V.nodes(),

        key=lambda x:
            degree[x],

        reverse=True

    )[:40]


    labels = {
        n: n
        for n in label_nodes
    }


    nx.draw_networkx_labels(

        V,

        pos,

        labels=labels,

        font_size=8,

        font_weight="bold"
    )


    # Important edge labels
    edge_labels = {}

    edge_candidates = []

    for u, v, data in V.edges(
        data=True
    ):

        score = (
            degree[u]
            +
            degree[v]
        )

        edge_candidates.append(
            (
                score,
                u,
                v,
                data
            )
        )


    edge_candidates.sort(
        reverse=True
    )


    for (
        score,
        u,
        v,
        data
    ) in edge_candidates[:35]:

        relations = data.get(
            "relations",
            set()
        )

        if relations:

            edge_labels[
                (u, v)
            ] = ", ".join(
                list(relations)[:2]
            )


    nx.draw_networkx_edge_labels(

        V,

        pos,

        edge_labels=edge_labels,

        font_size=7
    )


    plt.title(

        "ECAP-RAG — Connected "
        "Entity-Anchored Knowledge Graph\n"
        f"{V.number_of_nodes()} nodes · "
        f"{V.number_of_edges()} reasoning edges",

        fontsize=20,

        fontweight="bold"
    )


    plt.axis(
        "off"
    )

    plt.tight_layout()

    plt.savefig(

        PLOT_PATH,

        dpi=350,

        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 19. FINAL
# ============================================================

print()
print("=" * 70)
print("KG BUILD FINISHED")
print("=" * 70)

print(
    "Full nodes:",
    builder.graph.number_of_nodes()
)

print(
    "Full reasoning edges:",
    len(
        list(
            builder.reasoning_edges()
        )
    )
)

print(
    "Claims:",
    len(
        builder.claims
    )
)

print(
    "Largest connected component:",
    (
        len(components[0])
        if components
        else 0
    )
)

print()
print(
    "IMPORTANT:"
)

print(
    "If largest component is still only "
    "5–10 nodes, we should improve entity "
    "linking before implementing path retrieval."
)
```

```python
# ============================================================
# DRAW THE ENTIRE KG — NO FILTERING
# ============================================================

G_full = nx.Graph()

for (
    u,
    v,
    key,
    data
) in builder.reasoning_edges():

    relation = data.get(
        "relation",
        ""
    )

    G_full.add_node(u)
    G_full.add_node(v)

    if G_full.has_edge(u, v):

        G_full[u][v][
            "relations"
        ].add(relation)

    else:

        G_full.add_edge(
            u,
            v,
            relations={relation}
        )


print(
    "Full graph:",
    G_full.number_of_nodes(),
    "nodes /",
    G_full.number_of_edges(),
    "edges"
)


# ------------------------------------------------------------
# Layout
# ------------------------------------------------------------

pos = nx.spring_layout(
    G_full,
    seed=42,
    k=0.4,
    iterations=300
)


plt.figure(
    figsize=(30, 25)
)

nx.draw_networkx_edges(
    G_full,
    pos,
    alpha=0.15,
    width=0.5
)

nx.draw_networkx_nodes(
    G_full,
    pos,
    node_size=20,
    alpha=0.8
)

plt.title(
    "ECAP-RAG — Complete Knowledge Graph",
    fontsize=24,
    fontweight="bold"
)

plt.axis("off")

plt.tight_layout()

plt.show()
```
