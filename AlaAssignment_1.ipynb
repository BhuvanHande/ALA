```python
%%writefile Vec.py

from __future__ import annotations
from typing import Self
import math


class Vec:
    def __init__(self, src=None) -> None:
        if src is None:
            self.elements = tuple()
        else:
            elements = tuple(src)

            for x in elements:
                if not isinstance(x, (int, float)):
                    raise TypeError(f"All elements must be numbers: {type(x)}")

            self.elements = elements

    def mean(self) -> float:
        if len(self.elements) == 0:
            raise RuntimeError("cannot calculate mean")

        return sum(self.elements) / len(self.elements)

    def demean(self) -> Self:
        mean_val = self.mean()

        return Vec(
            tuple(x - mean_val for x in self.elements)
        )

    def std(self) -> float:
        if len(self.elements) == 0:
            raise RuntimeError("cannot calculate standard deviation")

        demeaned_vec = self.demean()

        squared_deviations = tuple(
            x ** 2 for x in demeaned_vec.elements
        )

        return math.sqrt(
            sum(squared_deviations) / len(self.elements)
        )
```

```python
from Vec import Vec
import math


print("Running test cases...")


# -------------------------
# MEAN TESTS
# -------------------------

v = Vec([3, 8, 14])
assert round(v.mean(), 3) == 8.333

v = Vec([-9, -14, -3])
assert round(v.mean(), 3) == -8.667

v = Vec([5, 5, 5, 5])
assert v.mean() == 5

try:
    Vec().mean()
    assert False, "mean should raise RuntimeError for empty vector"
except RuntimeError:
    pass

print("Mean tests passed")


# -------------------------
# DEMEAN TESTS
# -------------------------

v = Vec([4, 17, 23])
result = v.demean()

assert [round(x, 3) for x in result.elements] == [
    -10.667, 2.333, 8.333
]

v = Vec([2, 6, 14])
result = v.demean()

assert [round(x, 3) for x in result.elements] == [
    -5.333, -1.333, 6.667
]

# Property: mean of demeaned vector is zero
v = Vec([10, 20, 30, 40])
demeaned = v.demean()

assert math.isclose(demeaned.mean(), 0.0, abs_tol=1e-10)

# Demean returns a new vector
v = Vec([1, 2, 3])
result = v.demean()

assert result is not v

try:
    Vec().demean()
    assert False, "demean should raise RuntimeError for empty vector"
except RuntimeError:
    pass

print("Demean tests passed")


# -------------------------
# STANDARD DEVIATION TESTS
# -------------------------

v = Vec([2, 8])
assert v.std() == 3.0

v = Vec([5, 5, 5, 5])
assert v.std() == 0.0

v = Vec([1, 2, 3, 4, 5])
assert v.std() >= 0

try:
    Vec().std()
    assert False, "std should raise RuntimeError for empty vector"
except RuntimeError:
    pass

print("Standard deviation tests passed")


print("\nAll assignment tests passed!")
```
